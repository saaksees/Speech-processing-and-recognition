# Lab 8 – Hidden Markov Model (HMM)
Simulation of phoneme transitions for the word 'happy'

```python
import numpy as np
import pandas as pd
```

```python
# HMM for the word "happy" (/h/ /æ/ /p/ /i/)
states = ['/h/', '/æ/', '/p/', '/i/']
observations = ['Energy', 'Pitch', 'Duration']

# start at /h/
init_prob = np.array([1.0, 0.0, 0.0, 0.0])

# transitions tuned to follow the sequence /h/ -> /æ/ -> /p/ -> /i/
# each row must sum to 1
trans_probs = np.array([
    [0.00, 0.99, 0.01, 0.00],  # from /h/ mostly -> /æ/
    [0.00, 0.00, 0.99, 0.01],  # from /æ/ mostly -> /p/
    [0.00, 0.00, 0.00, 1.00],  # from /p/ -> /i/
    [0.00, 0.00, 0.00, 1.00]   # /i/ is absorbing (stay)
])

# emission probabilities (Energy, Pitch, Duration) per phoneme
# rows must sum to 1
emiss_probs = np.array([
    [0.6, 0.3, 0.1],  # /h/  — moderate energy, low duration
    [0.5, 0.3, 0.2],  # /æ/  — vowel: decent energy, a bit longer
    [0.7, 0.1, 0.2],  # /p/  — plosive: high energy burst, short
    [0.4, 0.5, 0.1]   # /i/  — high pitch characteristic vowel
])
```

```python
# Display HMM matrices
print('Initial Probabilities:')
display(pd.DataFrame([init_prob], columns=states))
print('\nTransition Probabilities:')
display(pd.DataFrame(trans_probs, columns=states, index=states))
print('\nEmission Probabilities:')
display(pd.DataFrame(emiss_probs, columns=observations, index=states))
```

```python
# Generate a single phoneme and observation sequence for "happy"
def generate_sequence(steps=4, seed=None):
    if seed is not None:
        np.random.seed(seed)
    seq = []
    obs_seq = []
    curr_state = np.random.choice(len(states), p=init_prob)  # should be 0 (/h/)
    for step in range(steps):
        seq.append(states[curr_state])
        emis = np.random.choice(observations, p=emiss_probs[curr_state])
        obs_seq.append(emis)
        if step < steps - 1:
            curr_state = np.random.choice(len(states), p=trans_probs[curr_state])
    return seq, obs_seq
```

```python
# Example run (set seed for reproducibility if you like)
seq, obs_seq = generate_sequence(seed=42)
print('Generated Phoneme Sequence:', seq)
print('Generated Observation Sequence:', obs_seq)
```

# Inference

The HMM correctly models the phoneme order of the word “happy” by generating the sequence /h/ → /æ/ → /p/ → /i/ exactly as expected from the transition structure. The corresponding observation sequence varies stochastically according to the emission probabilities, showing how acoustic features like Energy, Pitch, and Duration are probabilistically linked to each phoneme. This demonstrates that the model successfully captures both the temporal structure of speech and the feature variability associated with each sound.
